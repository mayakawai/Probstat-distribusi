```python
p_spam = 0.05
p_normal = 0.95

p_deteksi_spam = 0.90
p_salah_deteksi = 0.10

p_spam_terdeteksi = (
    p_deteksi_spam * p_spam
) / (
    (p_deteksi_spam * p_spam) +
    (p_salah_deteksi * p_normal)
)

print("Probabilitas Email Benar-benar Spam:",
      round(p_spam_terdeteksi, 4))

print("Persentase:",
      round(p_spam_terdeteksi * 100, 2), "%")
```

```python
from scipy.stats import bernoulli
import numpy as np
import matplotlib.pyplot as plt

p = 0.8

dist = bernoulli(p)

prob_berhasil = dist.pmf(1)
prob_gagal = dist.pmf(0)

mean = dist.mean()
variance = dist.var()

print("Probabilitas Berhasil:", prob_berhasil)
print("Probabilitas Gagal:", prob_gagal)
print("Nilai Harapan:", mean)
print("Varians:", variance)

# Visualisasi
x = [0, 1]
y = dist.pmf(x)

plt.bar(x, y, edgecolor='black')
plt.title("Distribusi Bernoulli")
plt.xlabel("Hasil Login (0=Gagal, 1=Berhasil)")
plt.ylabel("Probabilitas")
plt.xticks(x)
plt.show()
```

```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt

n = 10
p = 0.8

dist = binom(n, p)

probabilitas = dist.pmf(8)
mean = dist.mean()
variance = dist.var()

print("Probabilitas Tepat 8 Berhasil:", probabilitas)
print("Nilai Harapan:", mean)
print("Varians:", variance)

# Visualisasi
x = np.arange(0, n + 1)
y = dist.pmf(x)

plt.bar(x, y, edgecolor='black')
plt.title("Distribusi Binomial")
plt.xlabel("Jumlah Login Berhasil")
plt.ylabel("Probabilitas")
plt.xticks(x)
plt.show()
```

```python
from scipy.stats import poisson
import numpy as np
import matplotlib.pyplot as plt

lam = 4

dist = poisson(lam)

probabilitas = dist.pmf(6)
mean = dist.mean()
variance = dist.var()

print("Probabilitas Tepat 6 Request:", probabilitas)
print("Nilai Harapan:", mean)
print("Varians:", variance)

# Visualisasi
x = np.arange(0, 13)
y = dist.pmf(x)

plt.bar(x, y, edgecolor='black')
plt.title("Distribusi Poisson")
plt.xlabel("Jumlah Request Server")
plt.ylabel("Probabilitas")
plt.show()
```
